We are trying to model Brownian Motion which is the irrgular movement of particles suspended in a fluid which allowed Einstein to postulate the reality of the existence of the atom. We will do this by first building a simulation where a large hard disk interacts elastically with a gas made of small disks in 2D motion. We will increase the complexity of the project goal/outcome if this proves to be too simple (Maybe by modeling it in 3 dimensions or adding some other component like Langevin Equation to see where it breaks down etc.). To start, we need to keep track of the velocity, position and momentum of every disc. The computational methods that we have vaguely in mind so far are some sort of visualization and a type of Monte Carlo method if needed.

```python
import numpy as np

#Let's start with the interaction of two disks. One large. One small. Neither has acceleration so each has a constant velocity. We create a class
#Class for particles

class Particle:
    #Initilize 
    def __init__(self,name,x_position,y_position,mass,radius,x_velocity):
        self.name = name
        self.x_position = x_position
        self.y_position = y_position
        self.mass = mass #g
        self.radius = radius #cm
        self.x_velocity = x_velocity #cm/s
        
    def step(self):
        self.x_position += self.x_velocity 
        # self.y_position += self.y_velocity 
        
        
Particle1 = Particle("Bob",0,0,100, 10, 1)

print(Particle1.x_position)
for i in range(0,5):
    Particle1.step()
    print(Particle1.x_position)
    
    
```
